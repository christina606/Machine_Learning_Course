imports :

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

get data :


```python
mall_customers =  pd.read_csv('data\Mall_Customers.csv')
mall_customers
```

```python
df = mall_customers.iloc[:,[3,4]].values
df
```

scaling


```python
scaler = StandardScaler()
x_sacled = scaler.fit_transform(df)
```

```python
x_sacled
```

K-Means

```python
kmeans = KMeans(n_clusters=5, init='k-means++', random_state=42)
y_kmeans = kmeans.fit_predict(x_sacled)
```

PCA

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(x_sacled)
```

```python
x_sacled
```

Visualization :

```python
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y_kmeans, cmap='plasma')
plt.title('Clusters Visualization using PCA')
plt.show()
```
