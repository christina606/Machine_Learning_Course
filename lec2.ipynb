LIST METHODS

```python
#append()
list = [1,2.3,"gt",False]
print(list)
list.append("99")
print(list)
list.append(["99","hi"])
print(list)

#extend()

list = [1,2.3,"gt",False]
print(list)
list.extend("99")
print(list)
list.extend(["99","hi"])
print(list)
#remove()
list = [1,2.3,"gt",False]
print(list)
list.remove("gt")
print(list)
#sort()#a_z = 97| A_Z=65
list = [1,2,11,10,3,4,8,9]
print(list)
list.sort()
print(list)
list.sort(reverse=True)
print(list)
#reverse
list = [1,2.3,"gt",False]
print(list)
list.reverse()
print(list)
#clear()
list = [1,2.3,"gt",False]
print(list)
list.clear()
print(list)
#copy()
list = [1,2.3,"gt",False]

l=list.copy()
print(l)
#count()


list = [1,2.3,1,"gt",False]
print(list)
list.count(1)
print(list)

#index()

list = [1,2.3,1,"gt",False]
print(list)
list.index("gt")
print(list)
#insert()
list = [1,2.3,1,"gt",False]
print(list)
list.insert(1,"gt")
print(list)

#pop()
list = [1,2.3,1,"gt",False]
print(list)
list.index(0)
print(list)
```
